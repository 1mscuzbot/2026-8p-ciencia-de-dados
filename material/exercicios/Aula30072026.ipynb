Exercicio 2: Criar um programa que imprima os N numeros e diga quais são Pares ou Ímpares.

```python
num = int(input("Digite o numero: "))
```

```python
for numero in range(0,num):
  if(numero % 2 == 0):
    print(f"{numero} é par!")
  else:
    print(f"{numero} é impar!")
```
